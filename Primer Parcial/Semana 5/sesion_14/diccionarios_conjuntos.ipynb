## 1. Agenda de contactos usando dict: agregar, buscar, actualizar y eliminar

```python
contactos = {}

def agregar_contacto(nombre, telefono):
    contactos[nombre] = telefono
    print(f"Contacto {nombre} agregado.")

def buscar_contacto(nombre):
    if nombre in contactos:
        print(f"Teléfono de {nombre}: {contactos[nombre]}")
    else:
        print(f"Contacto {nombre} no encontrado.")

def actualizar_contacto(nombre, nuevo_telefono):
    if nombre in contactos:
        contactos[nombre] = nuevo_telefono
        print(f"Contacto {nombre} actualizado.")
    else:
        print(f"Contacto {nombre} no encontrado.")

def eliminar_contacto(nombre):
    if nombre in contactos:
        del contactos[nombre]
        print(f"Contacto {nombre} eliminado.")
    else:
        print(f"Contacto {nombre} no encontrado.")

def mostrar_contactos():
    if contactos:
        for nombre, telefono in contactos.items():
            print(f"{nombre}: {telefono}")
    else:
        print("No hay contactos en la agenda.")

# Ejemplo de uso
agregar_contacto("Alice", "123 456 7890")
agregar_contacto("Bob", "987 654 3210")
agregar_contacto("Charlie", "555 555 5555")
agregar_contacto("Diana", "444 444 4444")
agregar_contacto("Eve", "333 333 3333")
agregar_contacto("Frank", "222 222 2222")
print("\nEstos son los contactos actuales:\n")
mostrar_contactos()
print("\nVamos a buscar el contacto de Charlie:\n")
buscar_contacto("Charlie")
print("\nVamos a buscar un contacto que no exista:\n")
buscar_contacto("Byron")
print("\nVamos a actualizar el contacto de Alice:\n")
actualizar_contacto("Alice", "111 111 1111")
print("\nEstos son los contactos actuales:\n")
mostrar_contactos()
print("\nVamos a eliminar el contacto de Bob y Frank:\n")
eliminar_contacto("Bob")
eliminar_contacto("Frank")
print("\nEstos son los contactos actuales:\n")
mostrar_contactos()
```

## 2. Frecuencia de palabras en un texto (dict) y top 3 más comunes

En una lista podemos usar la funcion pata odernarla:

    sorted(iterable, key=None, reverse=False)


```python
texto = ("Python es un lenguaje de programación popular. Python es fácil de aprender. Python es el mejor lenguaje para Ciencia de Datos es así\n")
print(texto)

texto = texto.lower().split()
frecuencia_palabras = {}

for palabra in texto:
    palabra = palabra.strip(".,!?\"'()[]{};:")
    if palabra:
        frecuencia_palabras[palabra] = frecuencia_palabras.get(palabra, 0) + 1

print(frecuencia_palabras)
print("\n")

top_3 = sorted(frecuencia_palabras.items(), key=lambda x: x[1], reverse=True)[:3]

print(top_3)
print("\n")

print("Las 3 palabras más frecuentes son:")
for palabra, frecuencia in top_3:
    print(f"{palabra}: {frecuencia} veces") 
```

## 3. Alumnos inscritos en dos materias (sets): obtener inserccón, union y diferencia

```python
Calculo = {"Byron", "Leonardo", "Melanie", "Melina", "Pablo", "David", "Yariel", "Lyan"}
Finanzas = {"Camila", "Roberto", "Melanie", "David", "Arturo", "Lyan", "Rocio", "Leonardo"}

print("Insercción:", Calculo & Finanzas)
print("Union:", Calculo | Finanzas)
print("Diferencia a-b:", Calculo - Finanzas)
print("Direfencia simétrica:", Calculo ^ Finanzas)
```

## 5. Validar una lista de correos: usa un set para detectar duplicados y un dict para marcar valido/invalidos (siempre contiene ´@´ y ´.´)

```python
# Listado de correos electrónicos y nombres (con duplicados)
correos = {"ayala@gmail.com", "Byron", "melanie@gmail.com", "pablo@gmail.com", "Pablo", "Landan", "landan@gmail.com", "Byron.", "Leonardo@"}

# Diccionario para marcar correos válidos e inválidos
validacion_correos = {}
for correo in correos:
    if "@" in correo and "." in correo.split()[-1]: # Verifica que el correo tenga un formato básico @ y .
        validacion_correos[correo] = "Válido"
    else:
        validacion_correos[correo] = "Inválido"

print(validacion_correos)
```
