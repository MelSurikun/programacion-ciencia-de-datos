##### **🔹 Ejercicio 1:** Suma de elementos de una lista

```python
import random

tamaño_lista = 8

rango_min = random.randint(1, 30)
rango_max = random.randint(50, 100)

lista_aleatoria = [random.randint(rango_min, rango_max) for _ in range(tamaño_lista)]
suma_total = sum(lista_aleatoria)

print(f"Lista con {tamaño_lista} elementos entre {rango_min} y {rango_max}:\n")
print("- Lista:", lista_aleatoria)
print("- Suma total:", suma_total)
```

##### **🔹 Ejercicio 2:** Eliminar duplicados de una lista

```python
lista_con_duplicados = [1, 2, 2, 3, 4, 4, 5, 1, 6, 7, 7, 8, 9, 9, 9, 10, 11, 11, 11, 12]

lista_sin_duplicados = []


for elemento in lista_con_duplicados:
    if elemento not in lista_sin_duplicados:
        lista_sin_duplicados.append(elemento)

print("Lista original con duplicados:", lista_con_duplicados)
print("")
print("Lista sin duplicados:", lista_sin_duplicados)
```

##### **🔹 Ejercicio 3:** Combinar dos tuplas

```python
tupla1 = (11, 22, 33, 44, 55)
tupla2 = ("tarea", "sesion", "trece")
tupla3 = (True, False)

tupla_combinada = sum((tupla1, tupla2, tupla3), ())

print("Tupla de numeros:", tupla1)
print("Tupla de palabras:", tupla2)
print("Tupla de booleanos:", tupla3)
print("\nTupla combinada:", tupla_combinada)
```

##### **🔹 Ejercicio 4:** Encontrar elementos comunes entre dos listas

```python
lista1 = [11, "hola", True, "soy", "mel", 3.14]
lista2 = ["mel", 11, 3.14, False, "concierto", True]

comunes = []

for elemento in lista1:
    if elemento in lista2:
        comunes.append(elemento)

print("Lista 1:", lista1)
print("Lista 2:", lista2)
print("\nElementos comunes:", comunes)
```

##### **🔹 Ejercicio 5:** Convertir una lista en tupla y viceversa

```python
print("---> LISTA A TUPLA")
print("-------------------------------------------------------------\n")

lista_original = [1, 2, 3, 4, 5]
print("LISTA ORIGINAL:")
print("Contenido:", lista_original)
print("Tipo:", type(lista_original))

print("\n-------------------------------------------------------------\n")
print("OPERACIONES PERMITIDAS EN LISTAS:")
lista_original.append(6)  # Agregar elemento
print(" - Append: Agregar elemento ->", lista_original)

lista_original[0] = 100  # Modificar elemento
print(" - Modificar: Cambiar elemento ->", lista_original)

lista_original.remove(3)  # Eliminar elemento
print(" - Remove: Eliminar elemento ->", lista_original)

lista_original.extend([7, 8])  # Extender lista
print(" - Extend: Agregar múltiples ->", lista_original)

print("\n-------------------------------------------------------------\n")
print("" + "CONVERTIR LISTA A TUPLA")
tupla_convertida = tuple(lista_original)
print("\n- TUPLA CONVERTIDA:")
print("Contenido:", tupla_convertida)
print("Tipo:", type(tupla_convertida))

print("\n-------------------------------------------------------------\n")

print("- OPERACIONES NO PERMITIDAS EN TUPLAS:")
try:
    tupla_convertida.append(9)  # Intentar agregar
    print(" - Append: ERROR - No se ejecutó")
except AttributeError as e:
    print(" - Append: NO PERMITIDO ->", e)

try:
    tupla_convertida[0] = 200  # Intentar modificar
    print(" - Modificar: ERROR - No se ejecutó")
except TypeError as e:
    print(" - Modificar: NO PERMITIDO ->", e)

try:
    tupla_convertida.remove(4)  # Intentar eliminar
    print(" - Remove: ERROR - No se ejecutó")
except AttributeError as e:
    print(" - Remove: NO PERMITIDO ->", e)

print("\n-------------------------------------------------------------\n")
print("- OPERACIONES PERMITIDAS EN TUPLAS:")
print(" - Acceder: tupla[1] =", tupla_convertida[1])
print(" - Rango: tupla[1:4] =", tupla_convertida[1:4])
print(" - Contar: tupla.count(100) =", tupla_convertida.count(100))
print(" - Índice: tupla.index(4) =", tupla_convertida.index(4))
```
