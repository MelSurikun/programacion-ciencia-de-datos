Crear un notebook llamado `registro_estudiantes_listas_tuplas.ipynb` con el desarrollo de un mini-proyecto que utilice **listas**, **diccionarios**, **funciones** y **tuplas** para gestionar un registro de estudiantes.  
El programa debe incluir validaciones, búsqueda de información, cálculos y presentación de resultados en formato tabular.

---

### Ejercicio: Registro de estudiantes

```python
from typing import List, Dict, Any, Tuple

# Tipo personalizado para el registro
Registro = List[Dict[str, Any]]

# Función para agregar un estudiante con validaciones
def agregar_estudiante(registro: Registro, estudiante: Dict[str, Any]) -> None:
    assert isinstance(estudiante.get("nombre"), str) and estudiante["nombre"], "Nombre inválido"
    assert isinstance(estudiante.get("edad"), int) and estudiante["edad"] > 0, "Edad inválida"
    assert isinstance(estudiante.get("carrera"), str) and estudiante["carrera"], "Carrera inválida"
    assert isinstance(estudiante.get("promedio"), (int, float)) and 0 <= estudiante["promedio"] <= 10, "Promedio inválido"
    registro.append(estudiante)

# Función para mostrar la lista de estudiantes en formato de tabla
def listar_estudiantes(registro: Registro) -> None:
    print(f"{'Nombre':<12} {'Edad':<4} {'Carrera':<12} {'Promedio':<8}")
    print("-"*42)
    for est in registro:
        print(f"{est['nombre']:<12} {est['edad']:<4} {est['carrera']:<12} {est['promedio']:<8.2f}")

# Función para buscar estudiantes por carrera
def buscar_por_carrera(registro: Registro, carrera: str) -> Registro:
    return [e for e in registro if e.get("carrera", "").lower() == carrera.lower()]

# Función para encontrar al estudiante con el mejor promedio
def mejor_promedio(registro: Registro) -> Tuple[str, float]:
    if not registro:
        return ("", 0.0)
    m = max(registro, key=lambda e: e.get("promedio", 0))
    return (m["nombre"], m["promedio"])

# Función para calcular el promedio general del grupo
def promedio_general(registro: Registro) -> float:
    if not registro:
        return 0.0
    total = sum(e["promedio"] for e in registro)
    return total / len(registro)

# Pruebas de ejecución
registro: Registro = []

# Agregar estudiantes
agregar_estudiante(registro, {"nombre": "Ana", "edad": 20, "carrera": "Datos", "promedio": 9.1})
agregar_estudiante(registro, {"nombre": "Luis", "edad": 22, "carrera": "IA", "promedio": 8.7})
agregar_estudiante(registro, {"nombre": "Marta", "edad": 21, "carrera": "Datos", "promedio": 9.5})
agregar_estudiante(registro, {"nombre": "Carlos", "edad": 23, "carrera": "IA", "promedio": 7.9})

# Mostrar todos los estudiantes
print("Lista completa de estudiantes:")
listar_estudiantes(registro)

# Buscar por carrera
print("\n\nEstudiantes de la carrera 'Datos':")
listar_estudiantes(buscar_por_carrera(registro, "Datos"))

# Mejor promedio
nombre, prom = mejor_promedio(registro)
print(f"\n\nEstudiante con mejor promedio: {nombre} ({prom:.2f})")

# Promedio general
print(f"\n\nPromedio general del grupo: {promedio_general(registro):.2f}")
```

### Conclusiones

**Estructuras utilizadas:**

- **Listas:** Para almacenar varios estudiantes en una colección ordenada y poder recorrerlos fácilmente.  
- **Diccionarios:** Para representar cada estudiante con sus atributos (“nombre”, “edad”, “carrera”, “promedio”).  
- **Funciones:** Para dividir el programa en partes lógicas y reutilizables, aplicando validaciones y cálculos específicos.  
- **Tuplas:** Para devolver más de un valor en una sola salida (por ejemplo, nombre y promedio del mejor estudiante).  

**Motivos:**

Estas estructuras se usaron porque permiten organizar los datos de forma clara y sencilla.  
Las listas hacen posible agregar o buscar estudiantes fácilmente, los diccionarios permiten manejar información con claves significativas, y las funciones ayudan a mantener el código limpio y ordenado.  
Las tuplas resultan útiles cuando se necesita retornar más de un dato sin usar estructuras más complejas.
