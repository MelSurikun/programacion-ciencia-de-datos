# Tarea: Gráficos básicos con Matplotlib

En este notebook se desarrollan tres gráficos propios utilizando las funcionalidades básicas de Matplotlib aprendidas en la sesión 37.

```python
import matplotlib.pyplot as plt
import random
```

## Gráfico 1: Evolución de calificaciones durante el semestre

Se presenta un gráfico de línea que muestra cómo han evolucionado las calificaciones a lo largo de seis evaluaciones parciales.

```python
parciales = [1, 2, 3, 4, 5, 6]
calificaciones = [7.5, 8.2, 7.9, 8.5, 8.0, 8.8]

print("Parciales:", parciales)
print("Calificaciones:", calificaciones)
print("Promedio:", sum(calificaciones)/len(calificaciones))
```

```python
plt.plot(parciales, calificaciones, marker="o", linestyle="-")
plt.title("Evolución de calificaciones por parcial")
plt.xlabel("Parcial")
plt.ylabel("Calificación")
plt.grid(True)
plt.show()
```

**Observaciones:** Las calificaciones muestran una tendencia general positiva, aunque con algunas variaciones. Por ejemplo, en el parcial 3 hubo una pequeña caída, pero después se recuperó. La mejora del primer al último parcial fue de 1.3 puntos.

## Gráfico 2: Comparación de lenguajes de programación más utilizados

Se muestra un gráfico de barras con los lenguajes de programación más utilizados en ciencia de datos.

```python
lenguajes = ["Python", "R", "SQL", "Julia", "Java"]
porcentaje_uso = [68, 24, 18, 6, 10]

for i in range(len(lenguajes)):
    print(f"{lenguajes[i]}: {porcentaje_uso[i]}%")
```

```python
plt.bar(lenguajes, porcentaje_uso, color="steelblue")
plt.title("Lenguajes de programación más utilizados en ciencia de datos")
plt.xlabel("Lenguaje")
plt.ylabel("Porcentaje de uso")
plt.show()
```

**Observaciones:** Python es claramente el lenguaje más utilizado con 68%, casi tres veces más que R. Los porcentajes no suman 100% porque muchos profesionales usan más de un lenguaje al mismo tiempo.

## Gráfico 3: Distribución de tiempos de respuesta en una aplicación

Se genera un histograma con 500 tiempos de respuesta simulados para analizar cómo se distribuyen estos valores.

```python
tiempos_respuesta = [random.randint(50, 500) for _ in range(500)]

print("Cantidad de datos:", len(tiempos_respuesta))
print("Valor mínimo:", min(tiempos_respuesta))
print("Valor máximo:", max(tiempos_respuesta))
print("Primeros 10 valores:", tiempos_respuesta[:10])
```

```python
plt.hist(tiempos_respuesta, bins=15, color="forestgreen", edgecolor="black")
plt.title("Distribución de tiempos de respuesta")
plt.xlabel("Tiempo (ms)")
plt.ylabel("Frecuencia")
plt.show()
```

**Observaciones:** Como los datos se generaron con números aleatorios uniformes, el histograma muestra una distribución bastante plana donde cada rango tiene frecuencias similares. En datos reales de una aplicación, normalmente se verían más respuestas rápidas y pocas lentas.

## Conclusiones

Se practicaron los tres tipos de gráficos básicos de Matplotlib:

- **Gráfico de línea:** Útil para ver cómo cambia algo a lo largo del tiempo.
- **Gráfico de barras:** Sirve para comparar cantidades entre categorías.
- **Histograma:** Permite ver cómo se distribuyen los datos numéricos.

Los elementos como títulos, etiquetas y cuadrículas ayudan a que los gráficos sean más fáciles de entender.
