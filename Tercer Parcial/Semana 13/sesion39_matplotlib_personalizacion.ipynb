# Tarea: Personalización de gráficos con Matplotlib

En este notebook se desarrollan tres gráficos personalizados utilizando colores, estilos de línea, marcadores, leyendas y subplots, aplicando los conceptos aprendidos en la sesión 38.

```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
```

## Gráfico 1: Comparación de crecimiento de usuarios en dos plataformas

Se compara el crecimiento mensual de usuarios en dos plataformas digitales, utilizando diferentes colores, estilos y marcadores para distinguir cada una.

```python
meses = np.arange(1, 13)
usuarios_plataforma_a = [1000, 1450, 2100, 2800, 4200, 5800, 7500, 9200, 12000, 15500, 19000, 23000]
usuarios_plataforma_b = [800, 1200, 1900, 2700, 4500, 6800, 9500, 12000, 15200, 19000, 23500, 29000]

print("Meses:", list(meses))
print("Plataforma A - Inicio:", usuarios_plataforma_a[0], "Final:", usuarios_plataforma_a[-1])
print("Plataforma B - Inicio:", usuarios_plataforma_b[0], "Final:", usuarios_plataforma_b[-1])
```

```python
plt.figure(figsize=(10, 5))

plt.plot(meses, usuarios_plataforma_a, label="Plataforma A", color="blue", linestyle="-", marker="o")
plt.plot(meses, usuarios_plataforma_b, label="Plataforma B", color="red", linestyle="--", marker="s")

plt.title("Crecimiento mensual de usuarios por plataforma")
plt.xlabel("Mes")
plt.ylabel("Número de usuarios")
plt.legend()
plt.grid(True)
plt.show()
```

**Observaciones:** Aunque la Plataforma A empezó con más usuarios, la Plataforma B creció más rápido y terminó superándola. Los diferentes estilos de línea y marcadores hacen que sea fácil distinguir cada plataforma en el gráfico.

## Gráfico 2: Evolución de métricas de rendimiento en un proyecto

Se grafican tres métricas de un proyecto de software a lo largo de ocho semanas, cada una con un estilo diferente.

```python
semanas = np.arange(1, 9)
velocidad = [15, 18, 20, 24, 26, 30, 33, 36]
calidad = [85, 86, 88, 87, 90, 91, 93, 94]
satisfaccion = [70, 73, 76, 80, 82, 86, 88, 91]

print("Semanas:", list(semanas))
print("Velocidad inicial:", velocidad[0], "- final:", velocidad[-1])
print("Calidad inicial:", calidad[0], "- final:", calidad[-1])
print("Satisfacción inicial:", satisfaccion[0], "- final:", satisfaccion[-1])
```

```python
plt.figure(figsize=(10, 6))

plt.plot(semanas, velocidad, label="Velocidad (puntos/semana)", color="green", linestyle="-", marker="o")
plt.plot(semanas, calidad, label="Calidad (%)", color="orange", linestyle="-.", marker="o")
plt.plot(semanas, satisfaccion, label="Satisfacción del cliente (%)", color="purple", linestyle=":", marker="o")

plt.title("Evolución de métricas de rendimiento del proyecto")
plt.xlabel("Semana")
plt.ylabel("Valor")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

**Observaciones:** Las tres métricas mejoraron durante el proyecto. La satisfacción del cliente fue la que más creció (de 70% a 91%). La leyenda en la esquina inferior derecha ayuda a identificar cada línea sin que estorbe los datos.

## Gráfico 3: Análisis comparativo mediante subplots

Se crea una figura con cuatro subplots en una cuadrícula 2x2, cada uno mostrando un aspecto diferente del consumo energético.

```python
horas = np.arange(0, 24)
consumo_residencial = 50 + 20 * np.sin(horas / 4) + np.random.normal(0, 3, 24)
consumo_industrial = 120 + 40 * np.sin((horas - 6) / 4) + np.random.normal(0, 5, 24)

dias_semana = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]
consumo_promedio = [850, 900, 880, 920, 950, 700, 650]

temperaturas = np.random.normal(22, 3, 100)

print("Datos preparados para los 4 subplots")
print("Temperatura promedio:", round(np.mean(temperaturas), 2), "°C")
```

```python
plt.figure(figsize=(12, 8))

# Subplot 1
plt.subplot(2, 2, 1)
plt.plot(horas, consumo_residencial, color="blue", marker="o")
plt.title("Consumo residencial por hora")
plt.xlabel("Hora del día")
plt.ylabel("Consumo (kWh)")
plt.grid(True)

# Subplot 2
plt.subplot(2, 2, 2)
plt.plot(horas, consumo_industrial, color="red", linestyle="--", marker="s")
plt.title("Consumo industrial por hora")
plt.xlabel("Hora del día")
plt.ylabel("Consumo (kWh)")
plt.grid(True)

# Subplot 3
plt.subplot(2, 2, 3)
plt.bar(dias_semana, consumo_promedio, color="green")
plt.title("Consumo promedio por día de la semana")
plt.xlabel("Día")
plt.ylabel("Consumo (kWh)")

# Subplot 4
plt.subplot(2, 2, 4)
plt.hist(temperaturas, bins=15, color="orange", edgecolor="black")
plt.title("Distribución de temperaturas")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Frecuencia")

plt.suptitle("Análisis del consumo energético y factores ambientales", fontsize=14)
plt.tight_layout()
plt.show()
```

**Observaciones:** Los subplots permiten mostrar varios gráficos relacionados en una sola figura. Se nota que el consumo industrial es mayor que el residencial, y que los fines de semana el consumo baja bastante. El `suptitle` sirve para poner un título general que aplique a todos los subplots.
