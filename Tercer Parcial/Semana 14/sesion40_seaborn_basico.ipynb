# Tarea: Seaborn Básico

En este notebook se presentan tres gráficos propios utilizando la biblioteca Seaborn, aplicando los conceptos de la sesión 40 sobre visualización básica.

## Importación de bibliotecas y carga de datos

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Carga del dataset de salarios en México
df = pd.read_csv("salarios_mexico.csv")
print("Dimensiones del dataset:", df.shape)
df.head()
```

```python
# Exploración básica
df.info()
```

```python
df.describe()
```

## Gráfico 1: Histograma de distribución de salarios

Se utiliza `histplot` para analizar cómo se distribuyen los salarios mensuales en la muestra. Se aplica el estilo `whitegrid` para facilitar la lectura de los valores.

```python
sns.set_style("whitegrid")

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="salario_mensual_mxn", bins=25, color="steelblue")
plt.title("Distribución de salarios mensuales")
plt.xlabel("Salario mensual (MXN)")
plt.ylabel("Frecuencia")
plt.show()
```

La distribución muestra una concentración de salarios en rangos bajos y medios, con una cola hacia valores más altos. Esta forma sugiere una distribución asimétrica positiva típica de datos salariales.

## Gráfico 2: Frecuencia de registros por sexo

Se utiliza `countplot` para visualizar cuántos registros corresponden a cada categoría de sexo. Se cambia el estilo a `darkgrid` para este gráfico.

```python
sns.set_style("darkgrid")

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="sexo", hue="sexo", palette="Set2")
plt.title("Cantidad de registros por sexo")
plt.xlabel("Sexo")
plt.ylabel("Cantidad")
plt.show()
```

El gráfico muestra la proporción de hombres y mujeres en el dataset. Se observa que la distribución entre ambas categorías es relativamente similar, aunque puede existir una ligera diferencia.

## Gráfico 3: Salario promedio por sexo

Se utiliza `barplot` para calcular y mostrar el promedio de salarios según el sexo. Se aplica el estilo `ticks` para una presentación más limpia.

```python
sns.set_style("ticks")

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="sexo", hue="sexo", y="salario_mensual_mxn", palette="pastel", errorbar="sd")
plt.title("Salario promedio mensual por sexo")
plt.xlabel("Sexo"),
plt.ylabel("Salario promedio (MXN)")
plt.show()
```

El barplot presenta el salario promedio para cada grupo, incluyendo barras de error que representan la desviación estándar. Las diferencias observadas pueden indicar disparidad salarial entre grupos.

## Conclusiones

En este notebook se aplicaron tres tipos de gráficos básicos de Seaborn: histplot para distribuciones numéricas, countplot para frecuencias de categorías, y barplot para agregaciones por grupo. Se utilizaron diferentes estilos de Seaborn (whitegrid, darkgrid, ticks) para variar la presentación visual de cada gráfico.
