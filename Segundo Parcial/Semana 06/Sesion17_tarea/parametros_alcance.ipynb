### 1. potencia(base, exponente=2)

```python
def potencia(base, exponente=2):
    """
    Calcula la potencia de un número.
    Parámetros:
      - base: número base.
      - exponente: potencia (valor por defecto = 2).
    """
    return base ** exponente

# Ejemplos
print(potencia(5))       # 25  (usa exponente por defecto)
print(potencia(2, 3))    # 8

# Pruebas
assert potencia(3) == 9
assert potencia(2, 4) == 16
```


### 3. formatear_nombre(nombre, apellido, mayus=False)

```python
def formatear_nombre(nombre, apellido, mayus=False):
    """
    Regresa el nombre completo formateado.
    Parámetros:
      - nombre: nombre del usuario.
      - apellido: apellido del usuario.
      - mayus: si True, devuelve todo en mayúsculas (por defecto False).
    """
    completo = f"{nombre} {apellido}"
    return completo.upper() if mayus else completo

# Ejemplos
print(formatear_nombre("Leo", "Galván"))             # "Leo Galván"
print(formatear_nombre("Leo", "Galván", True))       # "LEO GALVÁN"

# Pruebas
assert formatear_nombre("Ana", "López") == "Ana López"
assert formatear_nombre("Ana", "López", True) == "ANA LÓPEZ"
```

### 4. descuento(precio, porcentaje=0.10)

```python
def descuento(precio, porcentaje=0.10):
    """
    Calcula el precio con descuento.
    Parámetros:
      - precio: precio original (número).
      - porcentaje: fracción del descuento (por defecto 0.10 = 10%).
    Retorna el precio final después de aplicar el descuento.
    """
    return precio * (1 - porcentaje)

# Ejemplos
print(descuento(100))        # 90.0   (10% de descuento)
print(descuento(250, 0.20))  # 200.0  (20% de descuento)

# Pruebas simples
assert descuento(100) == 90
assert descuento(50, 0.0) == 50
assert descuento(200, 0.5) == 100
```

### 5. convertir_temp(valor, a='F')

```python
def convertir_temp(valor, a='F'):
    """
    Convierte una temperatura entre Celsius, Fahrenheit y Kelvin.
    Parámetros:
      - valor: número que representa la temperatura (en la unidad de entrada).
      - a: destino: 'F' para Fahrenheit, 'C' para Celsius, 'K' para Kelvin.
    Suposición: si se pide convertir *a* 'F' o 'K', la entrada se asume en Celsius.
    Retorna la temperatura convertida.
    """
    if a == 'F':
        # Celsius -> Fahrenheit
        return (valor * 9/5) + 32
    elif a == 'K':
        # Celsius -> Kelvin
        return valor + 273.15
    elif a == 'C':
        # ya está en Celsius
        return valor
    else:
        raise ValueError("Destino inválido. Usa 'C', 'F' o 'K'.")

# Ejemplos
print(convertir_temp(0, 'F'))   # 32.0
print(convertir_temp(100, 'K')) # 373.15
print(convertir_temp(25, 'C'))  # 25

# Pruebas simples
assert abs(convertir_temp(0, 'F') - 32.0) < 1e-9
assert abs(convertir_temp(100, 'K') - 373.15) < 1e-9
assert convertir_temp(-40, 'F') == -40  # punto donde C == F
```

📌 ¿Por qué NO usar valores mutables como valores por defecto?

En Python los valores por defecto se crean una sola vez, cuando se define la función.
Si usas algo mutable como una lista o diccionario, todos los llamados de la función comparten el mismo objeto, lo que causa errores difíciles de detectar.
