# Mini Proyecto: Funciones Modulares para Limpieza de Datos

```python
"""
Mini proyecto: Funciones modulares para limpieza y análisis de datos.
Incluye:
- Eliminación de duplicados
- Reemplazo de valores nulos
- Conversión de tipos
- Estadísticas básicas: media, mediana, moda, rango
"""

import csv
from statistics import mean, median, mode

# --------------------------
# Funciones de limpieza
# --------------------------

def eliminar_duplicados(datos):
    """
    Elimina filas duplicadas de una lista de diccionarios.

    Parámetros:
        datos (list[dict]): Lista de registros.

    Regresa:
        list[dict]: Lista sin duplicados.
    """
    lista_unica = []
    for fila in datos:
        if fila not in lista_unica:
            lista_unica.append(fila)
    return lista_unica


def reemplazar_nulos(datos, columna, valor_default):
    """
    Reemplaza valores nulos ('' o None) en una columna.

    Parámetros:
        datos (list[dict]): Lista de registros.
        columna (str): Nombre de la columna a limpiar.
        valor_default: Valor que sustituirá a los nulos.

    Regresa:
        list[dict]: Lista actualizada.
    """
    for fila in datos:
        if fila[columna] == "" or fila[columna] is None:
            fila[columna] = valor_default
    return datos


def convertir_tipo(datos, columna, tipo_func):
    """
    Convierte el tipo de una columna con una función (int, float, str, etc.)

    Parámetros:
        datos (list[dict])
        columna (str)
        tipo_func (function)

    Regresa:
        list[dict]
    """
    for fila in datos:
        try:
            fila[columna] = tipo_func(fila[columna])
        except:
            pass
    return datos


# --------------------------
# Funciones estadísticas
# --------------------------

def estadisticas_columna(datos, columna):
    """
    Calcula media, mediana, moda y rango de una columna numérica.

    Parámetros:
        datos (list[dict])
        columna (str)

    Regresa:
        dict
    """
    valores = [fila[columna] for fila in datos]

    return {
        "media": mean(valores),
        "mediana": median(valores),
        "moda": mode(valores),
        "rango": max(valores) - min(valores)
    }


# --------------------------
# Función principal de prueba
# --------------------------

if __name__ == "__main__":
    # Cargar el CSV
    datos = []
    with open("datos.csv", newline="", encoding="utf-8") as f:
        lector = csv.DictReader(f)
        datos = list(lector)

    print("Datos originales:")
    print(datos)

    # Limpiar datos
    datos = eliminar_duplicados(datos)
    datos = reemplazar_nulos(datos, "edad", 0)
    datos = convertir_tipo(datos, "edad", int)
    datos = convertir_tipo(datos, "calificacion", float)

    print("\nDatos limpios:")
    print(datos)

    # Estadísticas
    print("\nEstadísticas de 'calificacion':")
    print(estadisticas_columna(datos, "calificacion"))
```
