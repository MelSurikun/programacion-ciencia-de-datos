# 5 funciones diferentes (matemáticas, cadenas, listas)


### 1. Función matemática: cuadrado(n)


```python
def cuadrado(n):
    """Regresa el cuadrado de un número n."""
    return n * n

# Ejemplos
print(cuadrado(5))   # 25
print(cuadrado(-3))  # 9

# Pruebas
assert cuadrado(2) == 4
assert cuadrado(0) == 0
```

### 2. Función con cadenas: contar_vocales(texto)

```python
def contar_vocales(texto):
    """Cuenta cuántas vocales tiene un texto."""
    vocales = "aeiouAEIOU"
    return sum(1 for c in texto if c in vocales)

# Ejemplos
print(contar_vocales("hola"))     # 2
print(contar_vocales("ESCOM"))    # 2

# Pruebas
assert contar_vocales("aaa") == 3
assert contar_vocales("xyz") == 0
```

### 3. Función con listas: maximo_lista(nums)

```python
def maximo_lista(nums):
    """Regresa el número más grande de una lista."""
    return max(nums)

# Ejemplos
print(maximo_lista([1, 5, 2]))   # 5
print(maximo_lista([-10, -2]))   # -2

# Pruebas
assert maximo_lista([3, 4, 9]) == 9
assert maximo_lista([0]) == 0
```

### 4. Función lógica: es_par(n)

```python
def es_par(n):
    """Regresa True si n es par, False si es impar."""
    return n % 2 == 0

# Ejemplos
print(es_par(4))  # True
print(es_par(7))  # False

# Pruebas
assert es_par(10) == True
assert es_par(9) == False
```

### 5. Función propia: promedio(nums)

```python
def promedio(nums):
    """Regresa el promedio de una lista de números."""
    return sum(nums) / len(nums)

# Ejemplos
print(promedio([10, 20, 30]))  # 20.0
print(promedio([5, 5, 5]))     # 5.0

# Pruebas
assert promedio([1, 2, 3]) == 2
assert promedio([100]) == 100
```
