# Utilizando la programación POO crea 2 clases con al menos un metodo

La **POO** es un paradigma de programación basado en el uso de **clases** y **objetos**.

| Concepto | Descripción |
|-----------|-------------|
| **Clase** | Molde o plantilla que define cómo son los objetos. |
| **Objeto** | Instancia concreta de una clase. |
| **Atributos** | Variables que describen las propiedades del objeto. |
| **Métodos** | Funciones que definen el comportamiento del objeto. |

```python
class Soldado:
    def __init__(self, nombre, edad = 18, rango = "Recluta", salario = 14411.03):
        self.nombre = nombre
        self.edad = edad
        self.rango = rango
        self.salario= salario
    
    def mostrar_informacion(self):
        return f"Hola, mi nombre es {self.nombre} y tengo {self.edad} años. Mi rango actaual como soldado es {self.rango} y mi salario es de ${self.salario} pesos mensuales."

soldado1 = Soldado("Juan Perez", 25, "Cabo", 20000)
soldado2 = Soldado("Luis Gomez")

print(soldado1.mostrar_informacion())
print(soldado2.mostrar_informacion())
```

```python
class VehiculoMilitar:
    def __init__(self, tipo, capacidad = 4, armamento = "Ametralladora"):
        self.tipo = tipo
        self.capacidad = capacidad
        self.armamento = armamento
    
    def mostrar_detalles(self):
        return f"Tipo de vehículo: {self.tipo}, Capacidad: {self.capacidad} personas, Armamento: {self.armamento}."
    
vehiculo1 = VehiculoMilitar("Tanque", 5, "Cañón")
vehiculo2 = VehiculoMilitar("Transporte de tropas", 20)

print(vehiculo1.mostrar_detalles())
print(vehiculo2.mostrar_detalles())
```
