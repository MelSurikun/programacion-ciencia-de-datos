# 1. Ejemplos de lectura y escritura de archivos .txt

```python
# Crear un archivo de texto con nombres de alumnos
with open("alumnos.txt", "w") as f:
    f.write("Ana\n")
    f.write("Beatriz\n")
    f.write("Alberto\n")
    f.write("Marta\n")
    f.write("Andrés\n")
    f.write("Pedro\n")
    f.write("Andrea\n")
```

```python
# Leer el archivo txt
with open("alumnos.txt", "r") as f:
    nombres = f.read()
    print("Contenido del archivo 'alumnos.txt':")
    print(nombres)
```

```python
# Crea un archivo de texto con datos
with open("datos.txt", "w") as archivo:
    archivo.write("Línea 1: El Sol es una estrella.\n")
    archivo.write("Línea 2: La Luna es un satelite.\n")
    archivo.write("Línea 3: Tres es mejor que dos.\n")
```

```python
# Lectura de archivo de texto
with open("datos.txt", "r") as archivo:
    contenido = archivo.read()
    print("Contenido del archivo 'datos.txt':")
    print(contenido)
```

# 2. Ejemplos de lectura y escritura de archivos .cvs

```python
import csv

# Crear un archivo CSV con datos de alumnos
with open("alumnos.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Ana"])
    writer.writerow(["Beatriz"])
    writer.writerow(["Alberto"])
    writer.writerow(["Marta"])
    writer.writerow(["Andrés"])
    writer.writerow(["Pedro"])
    writer.writerow(["Andrea"])
```

```python
# Leer el archivo CSV
with open("alumnos.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for fila in reader:
        print(fila)
```

```python
import csv

# Crear el archivo CSV con datos
with open("datos.csv", "a", newline="", encoding="utf-8") as archivo:
    archivo = csv.writer(archivo)
    archivo.writerow(["Línea 1: El Sol es una estrella."])
    archivo.writerow(["Línea 2: La Luna es un satelite."])
    archivo.writerow(["Línea 3: Tres es mejor que dos."])
```

```python
# Leer el archivo CSV
with open("datos.csv", "r", newline="", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea)
```

```python
# Agregar una línea al CSV
with open("datos.csv", "a", newline="", encoding="utf-8") as archivo:
    archivo_writer = csv.writer(archivo)
    archivo_writer.writerow(["Línea 4: La Tierra es un planeta"])
    print("Se ha agregado una nueva línea al archivo 'datos.csv'.")
    
# Leer el archivo CSV actualizado
with open("datos.csv", "r", newline="", encoding="utf-8") as archivo:
    for linea in archivo:
        print(linea)
```

# 3. Ejercicio propio

```python
# Crea un archivo .txt con datos "crudos" separados por guiones
with open('empleados_raw.txt', 'w') as f:
    f.write("Juan-Desarrollador-50000\n")
    f.write("Maria-Diseñadora-45000\n")
    f.write("Pedro-Gerente-75000\n")
```

```python
import csv

# Lista para guardar los datos ya procesados
datos_empleados = []

# 1. Leer el archivo .txt
print("Leyendo y procesando datos...")
try:
    with open('empleados_raw.txt', 'r') as archivo_in:
        for linea in archivo_in:
            # 1. Elimina espacios y el salto de línea al final (.strip())
            # 2. Divide la línea usando el guion como separador (.split('-'))
            nombre, ocupacion, sueldo_str = [d.strip() for d in linea.strip().split('-')]
            
            # Convierte el sueldo a un número entero
            sueldo = int(sueldo_str)
            
            # Agrega la fila de datos procesados a la lista
            datos_empleados.append([nombre, ocupacion, sueldo])
            
except FileNotFoundError:
    print("Error: No se encontró 'empleados_raw.txt'. Asegúrate de ejecutar el Paso 1.")


# 2. Escribir al archivo CSV
if datos_empleados:
    print("Escribiendo datos en 'empleados_procesados.csv'...")
    
    with open('empleados_procesados.csv', 'w', newline='') as archivo_out:
        escritor_csv = csv.writer(archivo_out)
        
        # Escribe la cabecera (primera fila)
        escritor_csv.writerow(['Nombre', 'Ocupacion', 'Sueldo'])
        
        # Escribe todas las filas de datos
        escritor_csv.writerows(datos_empleados)

    print("Conversión completada. Se creó 'empleados_procesados.csv'.")
else:
     print("Proceso finalizado, no se encontraron datos para escribir.")
```
