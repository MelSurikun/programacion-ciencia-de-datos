# 3 programas con manejo de errores y excepciones

```python
nombre_archivo = 'datos_existentes.txt'

try:
    # Intenta crear el archivo para asegurar el éxito en la primera ejecución
    with open(nombre_archivo, 'w') as f:
        f.write("Prueba de lectura exitosa.")
        
    print(f"Intento de abrir: {nombre_archivo}")
    with open(nombre_archivo, 'r') as archivo:
        contenido = archivo.read()
        
except FileNotFoundError:
    print(f"Error: El archivo '{nombre_archivo}' no se encontró en la ruta.")
    
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
    
else:
    # Se ejecuta SI NO hubo FileNotFoundError ni otros errores
    print("Lectura exitosa.")
    print(f"Contenido leído: **{contenido.strip()}**")
    
finally:
    # Se ejecuta SIEMPRE
    print("\n--- Operación de archivo finalizada. ---")
```

```python
dividendo = 100
divisor = 0 # Cambia este valor a 5 para ver el bloque 'else' en acción

try:
    print(f"Intentando dividir {dividendo} / {divisor}...")
    resultado = dividendo / divisor
    
except ZeroDivisionError:
    print("Error: No se puede dividir por cero.")
    
except Exception as e:
    print(f"Ocurrió otro tipo de error: {e}")
    
else:
    # Se ejecuta solo si NO hubo errores
    print("División exitosa.")
    print(f"El resultado es: **{resultado}**")
    
finally:
    # Se ejecuta siempre
    del dividendo
    del divisor
    print("\n--- Variables de operación limpiadas. ---")
```

```python
entrada_usuario = "hola" # Reemplaza con "hola" o "42" para probar

try:
    print(f"Intentando convertir la entrada: '{entrada_usuario}'")
    numero = int(entrada_usuario)
    
except ValueError:
    print(f"Error de Valor: '{entrada_usuario}' no puede ser convertido a un número entero.")
    numero = None
    
else:
    # Se ejecuta si la conversión int() fue exitosa
    print("Conversión exitosa.")
    print(f"El tipo de dato final es: **{type(numero)}**")
    
finally:
    # Se ejecuta siempre
    print("\n--- Fin del proceso de validación de entrada. ---")
```
