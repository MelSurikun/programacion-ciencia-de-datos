# Manejo de valores nulos y duplicados
Crea **ejemplos de limpieza de datos reales**, donde:
- Identifiques valores nulos.  
- Apliques al menos dos estrategias de imputación (media, mediana, valor fijo).  
- Detectes y elimines duplicados donde sea necesario.

```python
import pandas as pd
import numpy as np

# Creamos un diccionario con datos "sucios" intencionalmente
data_sucia = {
    'ID_Cliente': [1, 2, 3, 4, 5, 3, 6, 7, 8, 4, 9, 10],  # IDs duplicados: 3 y 4
    'Nombre': ['Ana', 'Luis', 'Beatriz', 'Carlos', np.nan, 'Beatriz', 'Elena', 'Jorge', np.nan, 'Carlos', 'Sofia', 'Miguel'],
    'Edad': [25, np.nan, 30, 22, 28, 30, np.nan, 35, 29, 22, 27, np.nan],
    'Email': ['ana@test.com', 'luis@test.com', None, 'carlos@test.com', 'marta@test.com', None, 'elena@test.com', 'jorge@test.com', None, 'carlos@test.com', 'sofia@test.com', 'miguel@test.com'],
    'Compras': [150.50, 200.00, np.nan, 50.00, 0.0, np.nan, 120.00, np.nan, 300.50, 50.00, 80.0, 100.0]
}

# Convertimos a DataFrame
df = pd.DataFrame(data_sucia)

print("--- DataFrame Original (Sucio) ---")
print(df)
print("\n--- Información General ---")
print(df.info())
```

```python
# 1. Contar cuántos datos faltan por columna
print("--- Valores Nulos por Columna ---")
print(df.isnull().sum())

# 2. Contar cuántas filas están totalmente repetidas
print(f"\nFilas duplicadas: {df.duplicated().sum()}")
```

```python
# Eliminamos duplicados y actualizamos el DataFrame
df = df.drop_duplicates()

# Verificamos
print(f"Filas restantes después de quitar duplicados: {len(df)}")
# Deberían quedar 10 filas (se fueron los 2 duplicados)
```

```python
# Rellenar vaores numericos
# Calculamos el promedio de edad (solo de los que sí tienen dato)
promedio_edad = df['Edad'].mean()
print(f"Rellenando edades faltantes con: {promedio_edad:.1f} años")

# Rellenamos los huecos
df['Edad'] = df['Edad'].fillna(promedio_edad)
```

```python
# Rellenar valores categóricos
df['Nombre'] = df['Nombre'].fillna("Cliente Anónimo")
df['Email'] = df['Email'].fillna("no_email@disponible.com")
```

```python
# Rellenar Compras con Nan con 0.0
df['Compras'] = df['Compras'].fillna(0.0)
```

```python
print("\n--- DataFrame Limpio y Procesado ---")
print(df)
print("\n--- Verificación Final ---")
print(df.isnull().sum()) # Debería dar todo 0
```
