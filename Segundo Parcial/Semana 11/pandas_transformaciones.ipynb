# Transformación de datos con Pandas
Crea **al menos 3 ejemplos propios** de:
- Creación de nuevas columnas.  
- Transformaciones con `apply()` y funciones `lambda`.  
- Reemplazo de valores con `map()` o `replace()`.

```python
import pandas as pd

data_pelis = {
    'Titulo': ['The Matrix', 'Titanic', 'Shrek', 'Sharknado', 'Avengers', 'Inception'],
    'Genero': ['Sci-Fi', 'Romance', 'Animation', 'Horror', 'Action', 'Sci-Fi'],
    'Duracion_min': [136, 195, 90, 86, 143, 148],
    'Rating_IMDB': [8.7, 7.8, 7.9, 3.3, 8.0, 8.8],
    'Año': [1999, 1997, 2001, 2013, 2012, 2010]
}

df = pd.DataFrame(data_pelis)

print("--- DataFrame Original ---")
print(df)
```

```python
# A. Operación matemática simple
# Convertimos minutos a horas dividiendo entre 60
df['Duracion_Horas'] = df['Duracion_min'] / 60

# B. Operación con fechas
# Calculamos la antigüedad restando el año actual al año de estreno
anio_actual = 2024
df['Antiguedad_anios'] = anio_actual - df['Año']

print("\n--- 1. Nuevas Columnas (Horas y Antiguedad) ---")
print(df[['Titulo', 'Duracion_Horas', 'Antiguedad_anios']])
```

```python
# A. Lógica condicional (if/else)
# Si dura más de 120 min es 'Larga', si no, 'Normal'
df['Categoria_Duracion'] = df['Duracion_min'].apply(lambda x: 'Larga' if x > 120 else 'Normal')

# B. Manipulación de Texto
# Creamos un código uniendo las primeras 3 letras del título en mayúsculas con el año
df['Codigo_Catalogo'] = df.apply(lambda row: row['Titulo'][:3].upper() + "-" + str(row['Año']), axis=1)

print("\n--- 2. Apply y Lambda ---")
print(df[['Titulo', 'Categoria_Duracion', 'Codigo_Catalogo']])
```

```python
# A. Map con diccionario (Traducción)
# Creamos un diccionario traductor
diccionario_generos = {
    'Sci-Fi': 'Ciencia Ficción',
    'Romance': 'Romance',
    'Animation': 'Animación',
    'Horror': 'Terror',
    'Action': 'Acción'
}
# Mapeamos la columna Genero
df['Genero_Esp'] = df['Genero'].map(diccionario_generos)

# B. Replace (Corrección puntual)
# Digamos que 'Sharknado' es tan mala que queremos cambiar su rating 3.3 a 0.0
df['Rating_IMDB'] = df['Rating_IMDB'].replace(3.3, 0.0)

print("\n--- 3. Map y Replace ---")
print(df[['Titulo', 'Genero', 'Genero_Esp', 'Rating_IMDB']])
```
