# Filtrado avanzado de datos en Pandas

Crea *5 ejemplos* de filtrado avanzado

```python
import pandas as pd
import random
```

```python
# DataFrame de Pacientes (Hospital)
pacientes = pd.DataFrame({
    "Nombre": [f"Paciente {chr(65+i)}" for i in range(20)], # Genera Paciente A, B, C...
    "Edad": [random.randint(15, 60) for _ in range(20)],
    "Peso_kg": [random.randint(50, 100) for _ in range(20)],
    "Altura_m": [round(random.uniform(1.50, 1.90), 2) for _ in range(20)],
    "Tipo_Sangre": [random.choice(["O+", "A+", "B-", "AB+"]) for _ in range(20)]
})

pacientes
```

```python
miMascara = (pacientes["Edad"] > 30) | (pacientes["Tipo_Sangre"] == "AB+")
miMascara
```

```python
pacientes[miMascara]["Nombre"]
```

```python
pacientes["Peso_kg"] > 70
```

```python
# Filtrado Simple peso mayor a 70 kg
pacientes[pacientes["Peso_kg"] > 70]
```

```python
pacientes["Tipo_Sangre"] == "AB+"
```

```python
# Filtrado Simple tipo sangre AB+
pacientes[pacientes["Tipo_Sangre"] == "AB+"]
```

```python
# Filtrado multiple edad mayor a 18 y tipo de sangre O+
pacientes[(pacientes["Edad"] > 30) & (pacientes["Tipo_Sangre"] == "O+")]
```

```python
pacientes.loc[[0, 3, 5], ["Nombre", "Tipo_Sangre"]]
```

```python
pacientes.loc[[0, 3, 5, 10]]
```

```python
pacientes.loc[[i for i in range(3, -1, -2)]]
```
