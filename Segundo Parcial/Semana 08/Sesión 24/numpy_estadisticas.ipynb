# **Operaciones matemáticas y estadísticas básicas con NumPy**

Entrega un notebook llamado `numpy_estadisticas.ipynb` que incluya:
- 5 conjuntos de datos aleatorios (puedes variar tamaño y distribución) y sus estadísticas (media, mediana, std, min, max).
- Una matriz 5x5 con análisis por axis (suma, media, min, max por filas y columnas).
- Una simulación de 1,000 lanzamientos de dado y el porcentaje de aparición de cada cara.
- Una breve conclusión sobre cuándo conviene usar rand, randn o randint.

```python
import numpy as np

# Fijar semilla para resultados reproducibles
np.random.seed(42)

print("=" * 60)
print("ANÁLISIS ESTADÍSTICO CON NUMPY")
print("=" * 60)

print("\nSEMILLA FIJA: np.random.seed(42) para resultados reproducibles")

# 1. CONJUNTOS DE DATOS ALEATORIOS Y SUS ESTADÍSTICAS
print("\n" + "=" * 60)
print("1. 5 CONJUNTOS DE DATOS ALEATORIOS Y ESTADÍSTICAS")
print("=" * 60)

def calcular_estadisticas(datos, nombre):
    """Calcula y muestra estadísticas descriptivas de un conjunto de datos"""
    print(f"\n--- {nombre} ---")
    print(f"Datos: {datos}")
    print(f"Tamaño: {len(datos)} elementos")
    print(f"Media: {np.mean(datos):.2f}")
    print(f"Mediana: {np.median(datos):.2f}")
    print(f"Desviación estándar: {np.std(datos):.2f}")
    print(f"Varianza: {np.var(datos):.2f}")
    print(f"Mínimo: {np.min(datos):.2f}")
    print(f"Máximo: {np.max(datos):.2f}")
    print(f"Rango: {np.ptp(datos):.2f}")
    print(f"Percentil 25: {np.percentile(datos, 25):.2f}")
    print(f"Percentil 75: {np.percentile(datos, 75):.2f}")

# Conjunto 1: Distribución uniforme
datos_uniforme = np.random.rand(20) * 100
calcular_estadisticas(datos_uniforme, "Distribución Uniforme (0-100)")

# Conjunto 2: Distribución normal
datos_normal = np.random.randn(30) * 10 + 50
calcular_estadisticas(datos_normal, "Distribución Normal (μ=50, σ=10)")

# Conjunto 3: Enteros aleatorios
datos_enteros = np.random.randint(1, 101, 25)
calcular_estadisticas(datos_enteros, "Enteros Aleatorios (1-100)")

# Conjunto 4: Distribución exponencial
datos_exponencial = np.random.exponential(scale=2.0, size=15)
calcular_estadisticas(datos_exponencial, "Distribución Exponencial (λ=0.5)")

# Conjunto 5: Datos con outliers
datos_outliers = np.concatenate([
    np.random.normal(50, 5, 18),
    np.array([100, 150, 200, -50])
])
calcular_estadisticas(datos_outliers, "Datos con Outliers")

# 2. ANÁLISIS DE MATRIZ 5x5 POR AXIS
print("\n" + "=" * 60)
print("2. ANÁLISIS DE MATRIZ 5x5 POR AXIS")
print("=" * 60)

# Crear matriz 5x5 con números aleatorios
matriz = np.random.randint(1, 101, size=(5, 5))
print("Matriz 5x5:")
print(matriz)

def analizar_axis(matriz, axis, nombre):
    """Analiza una matriz a lo largo de un axis específico"""
    print(f"\n--- Análisis por {nombre} (axis={axis}) ---")
    print(f"Suma: {np.sum(matriz, axis=axis)}")
    
    # Para media y std, formateamos cada elemento del array
    media = np.mean(matriz, axis=axis)
    std = np.std(matriz, axis=axis)
    
    print("Media: ", end="")
    print([f"{x:.2f}" for x in media])
    
    print(f"Mediana: {np.median(matriz, axis=axis)}")
    
    print("Desviación estándar: ", end="")
    print([f"{x:.2f}" for x in std])
    
    print(f"Mínimo: {np.min(matriz, axis=axis)}")
    print(f"Máximo: {np.max(matriz, axis=axis)}")

# Análisis por filas (axis=1)
analizar_axis(matriz, axis=1, nombre="filas")

# Análisis por columnas (axis=0)
analizar_axis(matriz, axis=0, nombre="columnas")

# Análisis completo de la matriz
print(f"\n--- Estadísticas globales de la matriz ---")
print(f"Suma total: {np.sum(matriz)}")
print(f"Media global: {np.mean(matriz):.2f}")
print(f"Mediana global: {np.median(matriz)}")
print(f"Desviación estándar global: {np.std(matriz):.2f}")
print(f"Mínimo global: {np.min(matriz)}")
print(f"Máximo global: {np.max(matriz)}")

# Mostrar matriz con estadísticas por fila y columna
print(f"\n--- Resumen visual de la matriz ---")
print("Matriz original:")
for i, fila in enumerate(matriz):
    print(f"Fila {i+1}: {fila} | Suma: {np.sum(fila)}, Media: {np.mean(fila):.2f}")

print("\nResumen por columnas:")
for j in range(matriz.shape[1]):
    columna = matriz[:, j]
    print(f"Col {j+1}: Media: {np.mean(columna):.2f}, Min: {np.min(columna)}, Max: {np.max(columna)}")

# 3. SIMULACIÓN DE LANZAMIENTOS DE DADO
print("\n" + "=" * 60)
print("3. SIMULACIÓN DE 1,000 LANZAMIENTOS DE DADO")
print("=" * 60)

# Simular 1000 lanzamientos de un dado de 6 caras
lanzamientos = np.random.randint(1, 7, 1000)
print(f"Primeros 20 lanzamientos: {lanzamientos[:20]}...")

# Calcular frecuencia de cada cara
caras, conteos = np.unique(lanzamientos, return_counts=True)
porcentajes = (conteos / 1000) * 100

print("\nResultados de 1,000 lanzamientos:")
print("Cara | Conteo | Porcentaje")
print("-" * 25)
for cara, conteo, porcentaje in zip(caras, conteos, porcentajes):
    print(f"  {cara}  |   {conteo:3d}  |   {porcentaje:5.2f}%")

# Estadísticas de la simulación
print(f"\nEstadísticas de la simulación:")
print(f"Media de los resultados: {np.mean(lanzamientos):.2f}")
print(f"Desviación estándar: {np.std(lanzamientos):.2f}")
print(f"Resultado más frecuente: {caras[np.argmax(conteos)]} (apareció {np.max(conteos)} veces)")

# Verificar que la suma de porcentajes es 100%
print(f"Suma de porcentajes: {np.sum(porcentajes):.2f}%")

# Análisis adicional: frecuencia relativa acumulada
print(f"\nFrecuencia relativa acumulada:")
acumulado = 0
for cara, conteo, porcentaje in zip(caras, conteos, porcentajes):
    acumulado += porcentaje
    print(f"Cara {cara}: {acumulado:5.2f}%")

# 4. COMPARACIÓN: rand, randn, randint
print("\n" + "=" * 60)
print("4. COMPARACIÓN: rand vs randn vs randint")
print("=" * 60)

# Generar muestras de cada función
np.random.seed(123)

muestra_rand = np.random.rand(1000)
muestra_randn = np.random.randn(1000)
muestra_randint = np.random.randint(0, 10, 1000)

print("MUESTRAS GENERADAS (primeros 10 elementos):")
print(f"rand:   {np.array2string(muestra_rand[:10], precision=4, suppress_small=True)}")
print(f"randn:  {np.array2string(muestra_randn[:10], precision=4, suppress_small=True)}")
print(f"randint: {muestra_randint[:10]}")

print("\nESTADÍSTICAS COMPARATIVAS:")
print("Función  |   Media   |   Std    |   Mín   |   Máx   | Distribución")
print("-" * 65)
print(f"rand     | {np.mean(muestra_rand):8.4f} | {np.std(muestra_rand):7.4f} | {np.min(muestra_rand):7.4f} | {np.max(muestra_rand):7.4f} | Uniforme [0,1)")
print(f"randn    | {np.mean(muestra_randn):8.4f} | {np.std(muestra_randn):7.4f} | {np.min(muestra_randn):7.4f} | {np.max(muestra_randn):7.4f} | Normal estándar")
print(f"randint  | {np.mean(muestra_randint):8.4f} | {np.std(muestra_randint):7.4f} | {np.min(muestra_randint):7d} | {np.max(muestra_randint):7d} | Enteros discretos")

# Histogramas de las distribuciones
print("\nRANGOS DE VALORES GENERADOS:")
print(f"rand - entre {np.min(muestra_rand):.4f} y {np.max(muestra_rand):.4f}")
print(f"randn - entre {np.min(muestra_randn):.4f} y {np.max(muestra_randn):.4f}")
print(f"randint - entre {np.min(muestra_randint)} y {np.max(muestra_randint)}")

# 5. CONCLUSIÓN SOBRE CUÁNDO USAR CADA FUNCIÓN
print("\n" + "=" * 60)
print("5. CONCLUSIÓN: ¿CUÁNDO USAR rand, randn O randint?")
print("=" * 60)

print("""
RAND (np.random.rand):
- Genera números desde distribución uniforme continua en [0, 1)
- USAR CUANDO:
  * Necesitas probabilidades (valores entre 0 y 1)
  * Simulaciones que requieren distribución uniforme
  * Generar colores RGB aleatorios
  * Inicialización de pesos en redes neuronales
- EJEMPLO: np.random.rand(100) → 100 números entre 0 y 1

RANDN (np.random.randn):
- Genera números desde distribución normal estándar (μ=0, σ=1)
- USAR CUANDO:
  * Trabajas con datos que siguen distribución normal
  * Ruido gaussiano en procesamiento de señales
  * Errores de medición simulados
  * Finanzas (movimientos de precios)
- EJEMPLO: np.random.randn(100) → 100 números con media 0 y std 1

RANDINT (np.random.randint):
- Genera enteros aleatorios en un rango discreto [low, high)
- USAR CUANDO:
  * Simular lanzamientos de dados
  * Seleccionar índices aleatorios
  * Generar datos categóricos
  * Juegos y simulaciones discretas
- EJEMPLO: np.random.randint(1, 7, 100) → 100 lanzamientos de dado

RESUMEN:
- rand:    Para valores continuos entre 0 y 1 (distribución uniforme)
- randn:   Para valores con distribución normal (media 0, std 1)  
- randint: Para valores enteros en un rango específico
""")

# EJEMPLO PRÁCTICO ADICIONAL
print("\n" + "=" * 60)
print("EJEMPLO PRÁCTICO: APLICACIONES TÍPICAS")
print("=" * 60)

print("\n1. SIMULACIÓN DE MONEDA (randint es más apropiado):")
moneda_randint = np.where(np.random.randint(0, 2, 10) == 0, "Cara", "Cruz")
print(f"10 lanzamientos: {moneda_randint}")

print("\n2. SIMULACIÓN DE ALTURAS (randn para distribución normal):")
alturas = np.random.randn(5) * 10 + 170
print(f"Alturas simuladas (cm): {alturas}")

print("\n3. GENERACIÓN DE COLORES (rand para valores continuos):")
colores_rgb = np.random.rand(3, 3)
print(f"3 colores RGB aleatorios:")
for i, color in enumerate(colores_rgb):
    print(f"  Color {i+1}: R={color[0]:.3f}, G={color[1]:.3f}, B={color[2]:.3f}")

# VERIFICACIÓN DE REPRODUCIBILIDAD
print("\n" + "=" * 60)
print("VERIFICACIÓN DE REPRODUCIBILIDAD")
print("=" * 60)

print("Con la semilla fijada (np.random.seed(42)), estos resultados")
print("serán exactamente los mismos cada vez que ejecutes el código.")

# Demostración de reproducibilidad
np.random.seed(42)
primera_ejecucion = np.random.rand(5)
np.random.seed(42)
segunda_ejecucion = np.random.rand(5)

print(f"\nPrimera ejecución:  {primera_ejecucion}")
print(f"Segunda ejecución: {segunda_ejecucion}")
print(f"¿Son iguales? {np.array_equal(primera_ejecucion, segunda_ejecucion)}")

# RESUMEN FINAL DE ESTADÍSTICAS
print("\n" + "=" * 60)
print("RESUMEN FINAL")
print("=" * 60)

print("""
ESTADÍSTICAS CALCULADAS EN ESTE NOTEBOOK:
- Media, mediana, moda
- Desviación estándar y varianza
- Mínimo, máximo y rango
- Percentiles 25 y 75
- Análisis por axis (filas y columnas)
- Frecuencias y porcentajes
- Distribuciones de probabilidad

FUNCIONES NUMPY UTILIZADAS:
- np.mean(), np.median(), np.std(), np.var()
- np.min(), np.max(), np.ptp()
- np.percentile()
- np.sum() con axis
- np.unique() con return_counts
- np.random.rand(), np.random.randn(), np.random.randint()
""")
```
