# **Carga de datos desde archivos y primeras operaciones**

```python
import pandas as pd

df = pd.read_csv("iris.csv")

print("=== HEAD() ===")
print(df.head(), "\n")

print("=== INFO() ===")
print(df.info(), "\n")

print("=== DESCRIBE() ===")
print(df.describe(), "\n")

df_limpio = df[["sepal_length", "petal_length"]]

df_limpio.to_csv("iris_limpio.csv", index=False)

print("Archivo 'iris_limpio.csv' generado correctamente.")
```
