### 5 ejemplos propios de Series.

#### Ejemplo 1: Crear una Series de edades


```python
import pandas as pd

# Series básica
edades = pd.Series([18, 21, 30, 25, 40])
print("Ejemplo 1 - Series de edades:")
print(edades)
print()
```

#### Ejemplo 2: Series con índices personalizados

```python
temperaturas = pd.Series([22, 25, 20, 18], index=["Lunes", "Martes", "Miércoles", "Jueves"])
print("Ejemplo 2 - Series con índices personalizados:")
print(temperaturas)
print()
```

#### Ejemplo 3: Series a partir de un diccionario


```python
puntos_jugadores = pd.Series({"Leo": 15, "Ana": 20, "Luis": 13})
print("Ejemplo 3 - Series desde diccionario:")
print(puntos_jugadores)
print()
```

#### Ejemplo 4: Operaciones matemáticas en Series

```python
ventas = pd.Series([100, 150, 200, 250])
print("Ejemplo 4 - Ventas + 10%:")
print(ventas * 1.10)
print()
```

#### Ejemplo 5: Filtrar valores mayores

```python
calificaciones = pd.Series([5, 7, 9, 10, 6])
print("Ejemplo 5 - Calificaciones aprobatorias (>= 7):")
print(calificaciones[calificaciones >= 7])
print()
```
