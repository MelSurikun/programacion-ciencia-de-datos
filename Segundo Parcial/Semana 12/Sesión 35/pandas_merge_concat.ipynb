# Tarea: Operaciones con DataFrames en pandas 
 
Este notebook muestra ejemplos prácticos de:
1. `merge()` con diferentes tipos de join
2. `pd.concat()` para concatenación vertical y horizontal
3. `join()` con índices simples y con `set_index()`

 - Importar librerías y crear carpeta data

```python
import pandas as pd
import numpy as np
import os

# Crear carpeta data si no existe
if not os.path.exists("data"):
    os.makedirs("data")
    print("Carpeta 'data' creada")
else:
    print("Carpeta 'data' ya existe")

# Creación de DataFrames de ejemplo

# Crear DataFrame de estudiantes
estudiantes = pd.DataFrame({
    'estudiante_id': [101, 102, 103, 104, 105],
    'nombre': ['Ana', 'Luis', 'Carlos', 'María', 'Sofía'],
    'carrera': ['Computación', 'Matemáticas', 'Computación', 'Física', 'Matemáticas'],
    'promedio': [8.5, 7.8, 9.0, 8.2, 7.5]
})

print("DataFrame de Estudiantes:")
print(estudiantes)
print("\n" + "="*60)

# Crear DataFrame de cursos inscritos
cursos_inscritos = pd.DataFrame({
    'estudiante_id': [101, 102, 103, 104, 106, 107],
    'curso_id': ['C01', 'C02', 'C03', 'C01', 'C04', 'C05'],
    'calificacion': [85, 90, 88, 92, 78, 85],
    'semestre': ['2024-1', '2024-1', '2024-1', '2024-2', '2024-1', '2024-2']
})

print("DataFrame de Cursos Inscritos:")
print(cursos_inscritos)
print("\n" + "="*60)

# Crear DataFrame de información de cursos
info_cursos = pd.DataFrame({
    'curso_id': ['C01', 'C02', 'C03', 'C04', 'C06'],
    'nombre_curso': ['Programación', 'Cálculo', 'Álgebra', 'Física I', 'Química'],
    'creditos': [6, 8, 8, 7, 6],
    'profesor': ['Dr. Pérez', 'Dra. Gómez', 'Dr. López', 'Dra. Martínez', 'Dr. Ramírez']
})

print("DataFrame de Información de Cursos:")
print(info_cursos)
print("\n" + "="*60)

# %% [markdown]
# # 1. USO DE `merge()` CON DIFERENTES TIPOS DE JOIN

# %% [markdown]
# ## 1.1 INNER JOIN
# **Solo incluye registros que tienen coincidencia en AMBOS DataFrames**

# %%
print("MERGE con INNER JOIN (intersección):")
inner_merge = pd.merge(estudiantes, cursos_inscritos, 
                       on='estudiante_id', 
                       how='inner')
print(inner_merge)
print(f"\nFilas en estudiantes: {len(estudiantes)}")
print(f"Filas en cursos_inscritos: {len(cursos_inscritos)}")
print(f"Filas después de inner join: {len(inner_merge)}")
print(" Interpretación: Solo los estudiantes que están en AMBAS tablas aparecen")
print("   (Estudiantes 101-104 tienen cursos, 105 no, 106-107 no son estudiantes)")

# %% [markdown]
# ## 1.2 LEFT JOIN
# **Incluye TODOS los registros del DataFrame izquierdo, y los coincidentes del derecho**

# %%
print("MERGE con LEFT JOIN (todos los estudiantes):")
left_merge = pd.merge(estudiantes, cursos_inscritos, 
                      on='estudiante_id', 
                      how='left')
print(left_merge)
print(f"\n Filas después de left join: {len(left_merge)}")
print(" Interpretación: TODOS los estudiantes aparecen (incluso Sofía - ID 105)")
print("   Si no tienen cursos, las columnas de cursos_inscritos tienen NaN")
print("   Esto es útil para ver qué estudiantes NO han inscrito cursos")

# %% [markdown]
# ## 1.3 RIGHT JOIN
# **Incluye TODOS los registros del DataFrame derecho, y los coincidentes del izquierdo**

# %%
print("MERGE con RIGHT JOIN (todos los cursos inscritos):")
right_merge = pd.merge(estudiantes, cursos_inscritos, 
                       on='estudiante_id', 
                       how='right')
print(right_merge)
print(f"\n Filas después de right join: {len(right_merge)}")
print(" Interpretación: TODOS los cursos inscritos aparecen")
print("   Incluye cursos de estudiantes que no están en la tabla de estudiantes (IDs 106, 107)")
print("   Esto podría indicar un error en los datos o estudiantes no registrados")

# %% [markdown]
# ## 1.4 OUTER JOIN (FULL JOIN)
# **Incluye TODOS los registros de AMBOS DataFrames**

# %%
print(" MERGE con OUTER JOIN (unión completa):")
outer_merge = pd.merge(estudiantes, cursos_inscritos, 
                       on='estudiante_id', 
                       how='outer')
print(outer_merge)
print(f"\n Filas después de outer join: {len(outer_merge)}")
print(" Interpretación: Aparecen TODOS los estudiantes Y TODOS los cursos inscritos")
print("   Es la unión completa: estudiantes sin cursos y cursos sin estudiantes")
print("   Muestra información completa pero puede tener muchos NaN")

# ## 1.5 MERGE con múltiples columnas

# %%
# Crear DataFrames para merge complejo
ventas_2023 = pd.DataFrame({
    'mes': ['Enero', 'Febrero', 'Marzo', 'Abril'],
    'producto': ['Laptop', 'Tablet', 'Laptop', 'Celular'],
    'ventas': [120, 85, 110, 95]
})

ventas_2024 = pd.DataFrame({
    'mes': ['Enero', 'Febrero', 'Marzo', 'Mayo'],
    'producto': ['Laptop', 'Tablet', 'Celular', 'Tablet'],
    'ventas': [130, 90, 105, 88]
})

print(" MERGE con múltiples columnas clave:")
merge_multi = pd.merge(ventas_2023, ventas_2024, 
                       on=['mes', 'producto'], 
                       how='inner',
                       suffixes=('_2023', '_2024'))
print(merge_multi)
print("\n Interpretación: Solo productos en los mismos meses de ambos años")
print("   Permite comparar ventas del mismo producto en el mismo mes entre años")

# #  2. USO DE `pd.concat()` 

# ## 2.1 Concatenación VERTICAL (mismas columnas)
# **Apilar DataFrames uno sobre otro**

# Crear DataFrames con las mismas columnas (semestres diferentes)
estudiantes_sem1 = pd.DataFrame({
    'estudiante_id': [201, 202, 203],
    'nombre': ['Juan', 'Patricia', 'Roberto'],
    'promedio': [8.0, 9.2, 7.5],
    'semestre': ['2024-1', '2024-1', '2024-1']
})

estudiantes_sem2 = pd.DataFrame({
    'estudiante_id': [204, 205],
    'nombre': ['Laura', 'Daniel'],
    'promedio': [8.8, 7.9],
    'semestre': ['2024-2', '2024-2']
})

estudiantes_sem3 = pd.DataFrame({
    'estudiante_id': [206],
    'nombre': ['Gabriela'],
    'promedio': [9.5],
    'semestre': ['2024-3']
})

print("CONCATENACIÓN VERTICAL (apilar semestres):")
vertical_concat = pd.concat([estudiantes_sem1, estudiantes_sem2, estudiantes_sem3],
                           ignore_index=True)  # Reiniciar índices
print(vertical_concat)
print(f"\nTotal de estudiantes concatenados: {len(vertical_concat)}")
print("Interpretación: Se combinan registros de diferentes semestres")
print("   Útil para consolidar datos de diferentes periodos en una sola tabla")

# ## 2.2 Concatenación HORIZONTAL (más columnas)
# **Unir DataFrames lado a lado**

# Crear DataFrames con diferentes columnas (mismo índice)
info_personal = pd.DataFrame({
    'estudiante_id': [301, 302, 303, 304],
    'nombre': ['Elena', 'Fernando', 'Isabel', 'Javier'],
    'edad': [20, 21, 19, 22]
})

info_academica = pd.DataFrame({
    'estudiante_id': [301, 302, 303, 304],
    'carrera': ['Medicina', 'Derecho', 'Arquitectura', 'Ingeniería'],
    'promedio': [8.9, 7.8, 9.1, 8.5],
    'beca': ['Sí', 'No', 'Sí', 'Sí']
})

print(" CONCATENACIÓN HORIZONTAL (unir información):")
# Primero asegurar mismo orden
info_personal = info_personal.sort_values('estudiante_id').reset_index(drop=True)
info_academica = info_academica.sort_values('estudiante_id').reset_index(drop=True)

horizontal_concat = pd.concat([info_personal, info_academica.drop('estudiante_id', axis=1)], 
                             axis=1)  # axis=1 para horizontal
print(horizontal_concat)
print("\n Interpretación: Se unen diferentes tipos de información del mismo estudiante")
print("   Cada DataFrame aporta columnas distintas pero para los mismos registros")
print("   axis=1 concatena horizontalmente (columnas)")

# ## 2.3 Concat con diferentes índices

# %%
print(" CONCAT con diferentes índices y keys:")
# Crear DataFrames con índices diferentes
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']}, index=[0, 1])
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']}, index=[2, 3])

concat_con_keys = pd.concat([df1, df2], keys=['Grupo1', 'Grupo2'])
print(concat_con_keys)
print("\nNiveles del índice MultiIndex:")
print(concat_con_keys.index)
print("\n Interpretación: 'keys' crea un MultiIndex que identifica el origen de cada fila")
print("   Útil para mantener trazabilidad de dónde vienen los datos")

# #  3. USO DE `join()` 


# ## 3.1 Join con índices simples

# %%
# Crear DataFrames con índices comunes
productos = pd.DataFrame({
    'precio': [1500, 800, 1200, 500],
    'categoria': ['Electrónica', 'Electrónica', 'Hogar', 'Oficina']
}, index=['P001', 'P002', 'P003', 'P004'])

inventario = pd.DataFrame({
    'cantidad': [50, 120, 80, 200],
    'ubicacion': ['Almacén A', 'Almacén B', 'Almacén A', 'Almacén C']
}, index=['P001', 'P002', 'P003', 'P004'])

print(" JOIN con índices simples (mismo índice):")
join_simple = productos.join(inventario)
print(join_simple)
print("\n Interpretación: Se unen por índice automáticamente")
print("   Ambos DataFrames comparten el mismo índice (códigos de producto)")
print("   Es más simple que merge cuando ya tienes índices comunes")

# ## 3.2 Join después de `set_index()`
# **Convertir una columna a índice para hacer join**

# Crear DataFrames con columna común
empleados = pd.DataFrame({
    'empleado_id': [1, 2, 3, 4],
    'nombre': ['Carlos', 'Ana', 'Luis', 'Marta'],
    'departamento': ['Ventas', 'IT', 'Ventas', 'RH']
})

salarios = pd.DataFrame({
    'empleado_id': [1, 2, 3, 5],
    'salario': [25000, 30000, 22000, 28000],
    'bonus': [5000, 7000, 3000, 4000]
})

print(" JOIN después de set_index():")
# Convertir 'empleado_id' a índice en ambos DataFrames
empleados_idx = empleados.set_index('empleado_id')
salarios_idx = salarios.set_index('empleado_id')

join_con_setindex = empleados_idx.join(salarios_idx, how='left')
print(empleados_idx)
print("\n➕")
print(salarios_idx)
print("\n🟰 RESULTADO (left join):")
print(join_con_setindex)
print("\n Interpretación: Convertimos 'empleado_id' a índice para usar join()")
print("   Marta (ID 4) no tiene info de salario → NaN")
print("   Empleado ID 5 no está en empleados → no aparece (porque es left join)")

# ## 3.3 Join con diferentes tipos de join

# %%
print(" JOIN con different how parameters:")
print("\n1. Inner join (solo coincidencias):")
inner_join = empleados_idx.join(salarios_idx, how='inner')
print(inner_join)

print("\n2. Outer join (todos):")
outer_join = empleados_idx.join(salarios_idx, how='outer')
print(outer_join)

print("\n Interpretación: 'how' funciona igual que en merge()")
print("   inner: solo empleados con salario registrado")
print("   outer: todos los empleados Y todos los salarios")

# # COMPARACIÓN Y RESUMEN

# %%
print("="*70)
print(" RESUMEN DE OPERACIONES CON DATAFRAMES")
print("="*70)

print("\n MERGE():")
print("  • Para unir por COLUMNAS específicas (on='columna')")
print("  • Tipos: inner, left, right, outer")
print("  • Útil cuando los datos se relacionan por columnas de clave")

print("\n CONCAT():")
print("  • Para COMBINAR DataFrames (apilar o unir)")
print("  • axis=0 → Vertical (mismas columnas)")
print("  • axis=1 → Horizontal (más columnas, mismo índice)")
print("  • Útil para consolidar datos de diferentes fuentes")

print("\n JOIN():")
print("  • Para unir por ÍNDICES")
print("  • Requiere que los DataFrames tengan índices comunes")
print("  • Se puede usar set_index() para convertir columnas a índices")
print("  • Más limpio que merge cuando ya trabajas con índices")

# ##  EJEMPLO INTEGRADO: Sistema de Biblioteca

# Ejemplo integrado: Sistema de biblioteca
print("\n" + "="*70)
print(" EJEMPLO INTEGRADO: SISTEMA DE BIBLIOTECA")
print("="*70)

# DataFrames
libros = pd.DataFrame({
    'libro_id': ['L001', 'L002', 'L003', 'L004'],
    'titulo': ['Cien años de soledad', 'El principito', '1984', 'Don Quijote'],
    'autor': ['Gabriel García Márquez', 'Antoine de Saint-Exupéry', 
              'George Orwell', 'Miguel de Cervantes'],
    'categoria': ['Novela', 'Fábula', 'Distopía', 'Novela']
})

prestamos_2023 = pd.DataFrame({
    'prestamo_id': ['P2023-001', 'P2023-002', 'P2023-003'],
    'libro_id': ['L001', 'L002', 'L003'],
    'usuario_id': ['U101', 'U102', 'U101'],
    'fecha_prestamo': ['2023-03-15', '2023-06-20', '2023-11-05']
})

prestamos_2024 = pd.DataFrame({
    'prestamo_id': ['P2024-001', 'P2024-002', 'P2024-003'],
    'libro_id': ['L001', 'L004', 'L002'],
    'usuario_id': ['U103', 'U101', 'U104'],
    'fecha_prestamo': ['2024-01-10', '2024-02-15', '2024-03-22']
})

print("\n1. Libros disponibles:")
print(libros)

print("\n2. Concatenar préstamos de diferentes años:")
prestamos_total = pd.concat([prestamos_2023, prestamos_2024], ignore_index=True)
print(prestamos_total)

print("\n3. Merge para ver información completa de préstamos:")
prestamos_completo = pd.merge(prestamos_total, libros, on='libro_id', how='left')
print(prestamos_completo)

print("\n4. Libros NUNCA prestados (left join con filtro):")
libros_prestados = pd.merge(libros, prestamos_total, on='libro_id', how='left')
nunca_prestados = libros_prestados[libros_prestados['prestamo_id'].isna()]
print(nunca_prestados[['libro_id', 'titulo', 'autor']])

# #  GUARDAR RESULTADOS

# Guardar algunos DataFrames como CSV para referencia
# Nota: La carpeta 'data' ya fue creada al inicio
libros.to_csv('data/libros_biblioteca.csv', index=False)
prestamos_completo.to_csv('data/prestamos_completos.csv', index=False)

print("\nArchivos guardados en carpeta 'data':")
print("  • libros_biblioteca.csv")
print("  • prestamos_completos.csv")

# Verificar que los archivos se crearon
import os
if os.path.exists("data/libros_biblioteca.csv"):
    print(f"\n Archivo creado: data/libros_biblioteca.csv ({os.path.getsize('data/libros_biblioteca.csv')} bytes)")
if os.path.exists("data/prestamos_completos.csv"):
    print(f" Archivo creado: data/prestamos_completos.csv ({os.path.getsize('data/prestamos_completos.csv')} bytes)")
```
