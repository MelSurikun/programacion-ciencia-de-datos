```python
from faker import Faker
import random
import csv

faker = Faker()

# Lista de 100 canciones realistas de diferentes artistas (incluyendo los que mencionaste)
canciones = [
    "Heather", "Maniac", "Wish You Were Sober", "Comfort Crowd", "Checkmate",
    "The Story", "Maniac (Acoustic)", "The Cut That Always Bleeds",
    "Dákiti", "Yonaguni", "Callaita", "Safaera", "Titi Me Preguntó",
    "Me Porto Bonito", "Efecto", "Neverita", "Moscow Mule", "La Canción",
    "God's Menu", "Maniac", "Thunderous", "Back Door", "Case 143",
    "Christmas EveL", "Domino", "Venom", "Charmer",
    "Billie Jean", "Thriller", "Beat It", "Smooth Criminal", "Bad",
    "Black or White", "Man in the Mirror", "The Way You Make Me Feel",
    "Blinding Lights - The Weeknd", "Shape of You - Ed Sheeran",
    "As It Was - Harry Styles", "Stay - The Kid LAROI, Justin Bieber",
    "Bad Guy - Billie Eilish", "Levitating - Dua Lipa",
    "Watermelon Sugar - Harry Styles", "Good 4 U - Olivia Rodrigo",
    "Dynamite - BTS", "Butter - BTS", "Permission to Dance - BTS",
    "Kill Bill - SZA", "Anti-Hero - Taylor Swift", "Cruel Summer - Taylor Swift",
    "Flowers - Miley Cyrus", "Calm Down - Rema, Selena Gomez",
    "Seven - Jungkook ft. Latto", "Like Crazy - Jimin",
    "Cupid - FIFTY FIFTY", "Unholy - Sam Smith, Kim Petras",
    "I'm Good (Blue) - David Guetta, Bebe Rexha",
    "Last Night - Morgan Wallen", "Fast Car - Luke Combs",
    "vampire - Olivia Rodrigo", "What Was I Made For? - Billie Eilish",
    "Dance The Night - Dua Lipa", "Popular - The Weeknd, Madonna, Playboi Carti",
    "Paint The Town Red - Doja Cat", "greedy - Tate McRae",
    "Lovin On Me - Jack Harlow", "Yes, And? - Ariana Grande",
    "My Love Mine All Mine - Mitski", "Lose Control - Teddy Swims",
    "Beautiful Things - Benson Boone", "Espresso - Sabrina Carpenter",
    "A Bar Song (Tipsy) - Shaboozey", "I Can Do It With A Broken Heart - Taylor Swift",
    "Houdini - Dua Lipa", "Not Like Us - Kendrick Lamar",
    "Please Please Please - Sabrina Carpenter", "Too Sweet - Hozier",
    "Birds of a Feather - Billie Eilish", "Die With A Smile - Lady Gaga, Bruno Mars",
    "TEXAS HOLD 'EM - Beyoncé", "Saturn - SZA", "The Door - Teddy Swims",
    "Fortnight - Taylor Swift ft. Post Malone", "Down Bad - Taylor Swift",
    "So Long, London - Taylor Swift", "Who's Afraid of Little Old Me? - Taylor Swift",
    "I Look in People's Windows - Taylor Swift", "The Alchemy - Taylor Swift",
    "Clara Bow - Taylor Swift", "The Black Dog - Taylor Swift",
    "Chloe or Sam or Sophia or Marcus - Taylor Swift", "How Did It End? - Taylor Swift",
    "So High School - Taylor Swift", "I Hate It Here - Taylor Swift",
    "thanK you aIMee - Taylor Swift", "loml - Taylor Swift",
    "The Prophecy - Taylor Swift", "Cassandra - Taylor Swift",
    "Peter - Taylor Swift", "The Bolter - Taylor Swift",
    "Robin - Taylor Swift", "The Manuscript - Taylor Swift"
][:100]  # Aseguramos que sean exactamente 100 canciones

# Lista de meses
meses = [
    "Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
    "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"
]

# Años de reproducción (últimos 10 años)
años = list(range(2015, 2025))

# Función para generar número de reproducciones realista
def generar_reproducciones():
    # Algunas canciones son más populares que otras
    if random.random() < 0.2:  # 20% de canciones muy populares
        return random.randint(100000, 1000000)
    elif random.random() < 0.5:  # 30% de canciones populares
        return random.randint(10000, 99999)
    else:  # 50% de canciones normales
        return random.randint(100, 9999)

# Función para generar número de likes basado en reproducciones
def generar_likes(reproducciones):
    # Entre 1% y 20% de las reproducciones se convierten en likes
    porcentaje_like = random.uniform(0.01, 0.2)
    return int(reproducciones * porcentaje_like)

# Crear archivo CSV
n = 100  # Número de registros
with open("data/reproducciones_musicales.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow([
        "id_reproduccion", 
        "nombre_cancion", 
        "año_reproduccion", 
        "mes_reproduccion", 
        "num_reproducciones", 
        "num_likes"
    ])

    for i in range(n):
        cancion = random.choice(canciones)
        año = random.choice(años)
        mes = random.choice(meses)
        reproducciones = generar_reproducciones()
        likes = generar_likes(reproducciones)
        
        writer.writerow([
            faker.uuid4(),      # ID único
            cancion,            # Nombre de la canción
            año,                # Año de reproducción
            mes,                # Mes de reproducción
            reproducciones,     # Número de reproducciones
            likes               # Número de likes
        ])

print(f"Archivo reproducciones_musicales.csv generado con éxito con {n} registros")
print(f"Total de canciones únicas en el dataset: {len(canciones)}")
```
