# Tarea: Tablas Dinámicas con pivot_table()
 
Este notebook muestra ejemplos prácticos de:
1. **Ventas totales por región y mes** (aggfunc="sum")
2. **Promedio de ventas por producto y región** (aggfunc="mean")
3. **Múltiples funciones de agregación** ("sum", "mean", "max")
4. **Uso de margins=True** para totales
5. **Uso de fill_value** para valores vacíos

- Importar librerías

```python
import pandas as pd
import numpy as np
import random

print("Librerías importadas")
print(f"Versión de pandas: {pd.__version__}")

# Crear DataFrame de ventas ficticias
# Crear datos de ventas ficticios
np.random.seed(42)  # Para reproducibilidad
random.seed(42)

# Generar 100 registros de ventas
datos_ventas = {
    'fecha': pd.date_range(start='2024-01-01', periods=100, freq='D').tolist(),
    'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], 100),
    'producto': np.random.choice(['Laptop', 'Tablet', 'Celular', 'Monitor', 'Teclado'], 100),
    'categoria': np.random.choice(['Electrónica', 'Computación', 'Accesorios'], 100),
    'cantidad': np.random.randint(1, 10, 100),
    'precio_unitario': np.random.choice([500, 300, 800, 150, 50], 100),
    'vendedor': np.random.choice(['Ana', 'Carlos', 'María', 'Luis', 'Sofía'], 100)
}

# Crear DataFrame
ventas_df = pd.DataFrame(datos_ventas)

# Calcular venta total
ventas_df['venta_total'] = ventas_df['cantidad'] * ventas_df['precio_unitario']

# Extraer mes y día de la semana
ventas_df['mes'] = ventas_df['fecha'].dt.month_name()
ventas_df['dia_semana'] = ventas_df['fecha'].dt.day_name()

print("DATAFRAME DE VENTAS (primeras 10 filas):")
print(ventas_df.head(10))
print(f"\nTotal de registros: {len(ventas_df)}")
print(f"Rango de fechas: {ventas_df['fecha'].min()} a {ventas_df['fecha'].max()}")

# Mostrar información del DataFrame
print("\nINFORMACIÓN DEL DATAFRAME:")
print(ventas_df.info())

print("\nESTADÍSTICAS BÁSICAS:")
print(ventas_df.describe())

#EJEMPLO 1: Ventas totales por región y mes (aggfunc="sum")

print("="*70)
print("EJEMPLO 1: VENTAS TOTALES POR REGIÓN Y MES")
print("="*70)

print("\nDATAFRAME BASE (muestra):")
print(ventas_df[['region', 'mes', 'venta_total']].head())

# Crear tabla dinámica
pivot_1 = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='region',
    columns='mes',
    aggfunc='sum',
    fill_value=0  # Llenar valores NaN con 0
)

print("\nLLAMADA A pivot_table():")
print('''pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='region',
    columns='mes',
    aggfunc='sum',
    fill_value=0
)''')

print("\nRESULTADO - Ventas totales por región y mes:")
print(pivot_1)

print("\nINTERPRETACIÓN:")
print("• Esta tabla muestra el TOTAL de ventas ($) para cada combinación de región y mes.")
print("• Cada celda representa la suma total de ventas para esa región en ese mes.")
print("• Por ejemplo, en la región 'Este' en 'April' se vendieron $3,300.")
print("• Útil para identificar:")
print("  - Regiones con mayores ventas")
print("  - Meses con mejor desempeño")
print("  - Patrones estacionales por región")

# EJEMPLO 1.1: Con margins=True para totales

# %%
print("\n" + "="*70)
print("EJEMPLO 1.1: VENTAS TOTALES CON MARGINS=TRUE")
print("="*70)

# Crear tabla dinámica con margins
pivot_1_margins = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='region',
    columns='mes',
    aggfunc='sum',
    fill_value=0,
    margins=True,  # Agregar totales
    margins_name='TOTAL'  # Nombre para los totales
)

print(" LLAMADA (con margins=True):")
print('''pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='region',
    columns='mes',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='TOTAL'
)''')

print("\n RESULTADO - Con totales (margins):")
print(pivot_1_margins)

print("\n INTERPRETACIÓN:")
print("• margins=True agrega fila y columna 'TOTAL' con:")
print("  - TOTAL por región (última columna)")
print("  - TOTAL por mes (última fila)")
print("  - GRAN TOTAL (esquina inferior derecha)")
print("• Ejemplo: La región 'Norte' vendió $7,700 en total.")
print("• Ejemplo: En 'January' se vendieron $9,250 en total.")
print("• GRAN TOTAL: $41,100 en ventas en todo el período.")

# EJEMPLO 2: Promedio de ventas por producto y región (aggfunc="mean")

print("\n" + "="*70)
print(" EJEMPLO 2: PROMEDIO DE VENTAS POR PRODUCTO Y REGIÓN")
print("="*70)

print("\n DATAFRAME BASE (muestra):")
print(ventas_df[['producto', 'region', 'venta_total']].head())

# Crear tabla dinámica para promedio
pivot_2 = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='producto',
    columns='region',
    aggfunc='mean',
    fill_value='Sin ventas'  # Llenar valores NaN con texto
)

print("\n LLAMADA A pivot_table():")
print('''pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='producto',
    columns='region',
    aggfunc='mean',
    fill_value='Sin ventas'
)''')

print("\n RESULTADO - Promedio de ventas por producto y región:")
print(pivot_2)

print("\n INTERPRETACIÓN:")
print("• Esta tabla muestra el PROMEDIO de ventas por transacción.")
print("• Cada celda representa el valor promedio de una venta de ese producto en esa región.")
print("• Por ejemplo, en promedio:")
print("  - Las ventas de 'Celular' en la región 'Norte' son de $600")
print("  - Las ventas de 'Laptop' en la región 'Este' son de $800")
print("• 'Sin ventas' indica combinaciones sin transacciones.")
print("• Útil para identificar:")
print("  - Productos con mayor valor promedio por venta")
print("  - Diferencias regionales en el ticket promedio")
print("  - Oportunidades de ventas cruzadas")

#  EJEMPLO 2.1: Con margins=True y redondeo

print("\n" + "="*70)
print(" EJEMPLO 2.1: PROMEDIO CON MARGINS Y REDONDEO")
print("="*70)

# Crear tabla dinámica con márgenes
pivot_2_margins = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='producto',
    columns='region',
    aggfunc='mean',
    margins=True,
    margins_name='PROMEDIO TOTAL'
)

# Redondear a 2 decimales
pivot_2_margins_rounded = pivot_2_margins.round(2)

print("\n RESULTADO - Promedios con márgenes (redondeado):")
print(pivot_2_margins_rounded)

print("\n INTERPRETACIÓN:")
print("• Los márgenes muestran:")
print("  - PROMEDIO por producto (última columna)")
print("  - PROMEDIO por región (última fila)")
print("• Ejemplo: El promedio de ventas de 'Monitor' es $675.00")
print("• Ejemplo: El promedio de ventas en la región 'Sur' es $580.56")
print("• PROMEDIO TOTAL: $411.00 por transacción en general")

#  EJEMPLO 3: Múltiples funciones de agregación

print("\n" + "="*70)
print(" EJEMPLO 3: MÚLTIPLES FUNCIONES DE AGREGACIÓN")
print("="*70)

print("\n DATAFRAME BASE (muestra):")
print(ventas_df[['categoria', 'mes', 'venta_total', 'cantidad']].head())

# Crear tabla dinámica con múltiples funciones
pivot_3 = pd.pivot_table(
    data=ventas_df,
    values=['venta_total', 'cantidad'],  # Múltiples valores
    index='categoria',
    columns='mes',
    aggfunc={
        'venta_total': ['sum', 'mean', 'max'],  # Diferentes funciones para venta_total
        'cantidad': 'sum'  # Una función para cantidad
    },
    fill_value=0
)

print("\n LLAMADA A pivot_table() con múltiples funciones:")
print('''pd.pivot_table(
    data=ventas_df,
    values=['venta_total', 'cantidad'],
    index='categoria',
    columns='mes',
    aggfunc={
        'venta_total': ['sum', 'mean', 'max'],
        'cantidad': 'sum'
    },
    fill_value=0
)''')

print("\n RESULTADO - Análisis completo por categoría y mes:")
print(pivot_3)

print("\n INTERPRETACIÓN (niveles del MultiIndex):")
print("• Nivel 1: Variable ('venta_total' o 'cantidad')")
print("• Nivel 2: Función de agregación ('sum', 'mean', 'max')")
print("• Filas: Categorías de productos")
print("• Columnas: Meses")

print("\n LECTURA DE LA TABLA:")
print("• venta_total - sum: Total de ventas en $")
print("• venta_total - mean: Promedio de venta por transacción")
print("• venta_total - max: Venta individual más grande")
print("• cantidad - sum: Total de unidades vendidas")

print("\n EJEMPLOS:")
print("• 'Accesorios' en 'January':")
print("  - Total ventas: $1,950")
print("  - Promedio por venta: $325")
print("  - Venta máxima: $500")
print("  - Unidades vendidas: 15")

# EJEMPLO 4: Análisis avanzado de ventas

print("\n" + "="*70)
print(" EJEMPLO 4: ANÁLISIS AVANZADO - Vendedores por región y día")
print("="*70)

# Crear DataFrame adicional para análisis de vendedores
print("\n DATAFRAME BASE para análisis de vendedores:")
print(ventas_df[['vendedor', 'region', 'dia_semana', 'venta_total']].head())

# Tabla dinámica avanzada
pivot_4 = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index=['vendedor', 'region'],  # MultiIndex en filas
    columns='dia_semana',
    aggfunc=['count', 'sum', 'mean'],  # Múltiples funciones
    fill_value=0,
    margins=True,
    margins_name='TOTAL'
)

print("\n LLAMADA A pivot_table() avanzada:")
print('''pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index=['vendedor', 'region'],
    columns='dia_semana',
    aggfunc=['count', 'sum', 'mean'],
    fill_value=0,
    margins=True,
    margins_name='TOTAL'
)''')

print("\n RESULTADO - Desempeño de vendedores:")
print(pivot_4)

print("\n INTERPRETACIÓN (MultiIndex complejo):")
print("• Filas: Combinación vendedor + región")
print("• Columnas: Día de la semana")
print("• Valores: Tres métricas para cada combinación")
print("\n MÉTRICAS:")
print("• count: Número de ventas realizadas")
print("• sum: Total de ventas en $")
print("• mean: Promedio por venta")

print("\n EJEMPLO DE LECTURA:")
print("• Ana en la región Norte los Lunes:")
print("  - count: 2 ventas")
print("  - sum: $800 en total")
print("  - mean: $400 por venta (promedio)")

# EJEMPLO 5: Análisis de cantidad vs valor

print("\n" + "="*70)
print(" EJEMPLO 5: CANTIDAD VS VALOR - Doble perspectiva")
print("="*70)

# Tabla dinámica con análisis dual
pivot_5 = pd.pivot_table(
    data=ventas_df,
    values=['venta_total', 'cantidad'],
    index='producto',
    columns='region',
    aggfunc='sum',
    fill_value=0,
    margins=True
)

print("\n RESULTADO - Ventas totales y cantidad por producto y región:")
print(pivot_5)

print("\n INTERPRETACIÓN DUAL:")
print("• Para cada producto y región, vemos DOS perspectivas:")
print("  1. venta_total (sum): Total monetario en $")
print("  2. cantidad (sum): Número de unidades vendidas")
print("\n ANÁLISIS DE EJEMPLO:")
print("• 'Celular' en región 'Sur':")
print("  - venta_total: $3,200 (alto valor monetario)")
print("  - cantidad: 8 unidades (volumen moderado)")
print("  → Producto de alto valor, volumen medio")
print("\n• 'Teclado' en región 'Norte':")
print("  - venta_total: $450 (bajo valor monetario)")
print("  - cantidad: 9 unidades (alto volumen)")
print("  → Producto de bajo valor, alto volumen")

#  EJEMPLO 6: Formateo manual SIN .style (alternativa)

print("\n" + "="*70)
print(" EJEMPLO 6: TABLA FORMATEADA MANUALMENTE")
print("="*70)

# Crear tabla básica
pivot_6 = pd.pivot_table(
    data=ventas_df,
    values='venta_total',
    index='producto',
    columns='mes',
    aggfunc='sum',
    fill_value=0
)

print("\n TABLA ORIGINAL:")
print(pivot_6)

print("\n TABLA FORMATEADA MANUALMENTE (con formato monetario):")

# Crear una copia para formatear
pivot_6_formatted = pivot_6.copy()

# Aplicar formato monetario a todas las celdas
for col in pivot_6_formatted.columns:
    pivot_6_formatted[col] = pivot_6_formatted[col].apply(lambda x: f"${x:,.0f}")

print(pivot_6_formatted)

print("\nRESUMEN ESTADÍSTICO DE LA TABLA:")
print(f"• Valor máximo: ${pivot_6.max().max():,.0f}")
print(f"• Valor mínimo: ${pivot_6.min().min():,.0f}")
print(f"• Promedio: ${pivot_6.mean().mean():,.0f}")
print(f"• Total general: ${pivot_6.sum().sum():,.0f}")

print("\n TOP 3 COMBINACIONES PRODUCTO-MES:")
# Aplanar la tabla para análisis
flat_table = pivot_6.stack().reset_index()
flat_table.columns = ['Producto', 'Mes', 'Ventas']
top_3 = flat_table.sort_values('Ventas', ascending=False).head(3)

for i, (_, row) in enumerate(top_3.iterrows(), 1):
    print(f"{i}. {row['Producto']} en {row['Mes']}: ${row['Ventas']:,.0f}")

# RESUMEN Y COMPARACIÓN DE PARÁMETROS

print("\n" + "="*70)
print(" RESUMEN DE PARÁMETROS DE pivot_table()")
print("="*70)

print("\n PARÁMETROS ESENCIALES:")
print("1. data: DataFrame fuente (obligatorio)")
print("2. values: Columna(s) a agregar")
print("3. index: Columna(s) para las filas")
print("4. columns: Columna(s) para las columnas")
print("5. aggfunc: Función de agregación ('sum', 'mean', 'count', etc.)")
print("6. fill_value: Valor para reemplazar NaN (ej: 0 o 'Sin datos')")
print("7. margins: True/False para agregar totales")
print("8. margins_name: Nombre para la fila/columna de totales")

print("\n aggfunc MÁS COMUNES:")
aggfuncs = {
    'sum': 'Suma total',
    'mean': 'Promedio',
    'count': 'Conteo de registros',
    'min': 'Valor mínimo',
    'max': 'Valor máximo',
    'std': 'Desviación estándar',
    'var': 'Varianza',
    'first': 'Primer valor',
    'last': 'Último valor'
}

for func, desc in aggfuncs.items():
    print(f"  • {func}: {desc}")

print("\n USOS DE fill_value:")
print("  • fill_value=0: Para datos numéricos (evita NaN)")
print("  • fill_value='Sin datos': Para texto descriptivo")
print("  • fill_value=np.nan: Mantener NaN (por defecto)")

print("\n CUÁNDO USAR margins=True:")
print("  • Cuando necesitas totales por fila/columna")
print("  • Para resúmenes ejecutivos")
print("  • Para calcular porcentajes del total")
```
